```python
# functions + default args
def greet(name, excited=False):
    return f"Hello {name}!" if not excited else f"HELLO {name}!!"
```

```python
greet("Sharad")
greet("Sharad Singh", excited=True)

```

```python
# list comprehension
squares = [x**2 for x in range(10) if x % 2 == 0]
print(squares)
```

```python
# dict comprehension
word_lengths = {w: len(w) for w in ["ai", "machine", "learning"]}
print(word_lengths)
```

```python
# a simple class
class Movie:
    def __init__(self, title, rating):
        self.title = title
        self.rating = rating
    def is_hit(self):
        return self.rating >= 7.0
    def is_flop(self):
        return self.rating < 5.0

m = Movie("Inception", 8.8)
print(f"Is {m.title} a hit? {m.is_hit()}")
print(f"Is {m.title} a flop? {m.is_flop()}")
```
